# Random Forest Example 1: Random Feature Sampling

This notebook explains the **feature sampling** idea used in Random Forest.

In Random Forest, each decision tree does not always use all features.  
It randomly selects a subset of features so that trees become different from one another.

Educational dental example only — not for real diagnosis.

```
import pandas as pd

data = pd.read_csv("dental_random_forest_data.csv")
data
```

## Separate features and target

`X` contains input columns.  
`y` contains the output column that we want to predict.

```
X = data.drop("treatment", axis=1)
y = data["treatment"]

print("Features:")
print(X.head())

print("\nTarget:")
print(y.head())
```

## Randomly select some features

This simulates the Random Forest idea of selecting a random subset of columns/features.

```
random_features = X.sample(n=3, axis=1, random_state=42)

print("Randomly selected features:")
print(random_features.head())
```

## Train a Random Forest using feature randomness

`max_features=3` means each tree considers only 3 features at a split.

```
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)

model = RandomForestClassifier(
    n_estimators=100,
    max_features=3,
    random_state=42
)

model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Predictions:", y_pred)
print("Actual:", list(y_test))
print("Accuracy:", accuracy_score(y_test, y_pred))
```

## Feature importance

This shows which features were most useful for the forest.

```
importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
}).sort_values(by="Importance", ascending=False)

importance
```
