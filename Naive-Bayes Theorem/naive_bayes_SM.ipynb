# Naive Bayes Classifier — Play Tennis Dataset

This notebook implements a simple Naive Bayes classifier using the classic Play Tennis dataset.

Dataset columns:
- Outlook
- Temperature
- Humidity
- Wind
- PlayTennis

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.naive_bayes import CategoricalNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

## Step 1: Load Dataset

```python
df = pd.read_csv('play_tennis_video_dataset.csv')
df
```

## Step 2: Check Dataset Information

```python
print(df.shape)
print(df.info())
print(df['PlayTennis'].value_counts())
```

## Step 3: Separate Input Features and Target

```python
X = df.drop('PlayTennis', axis=1)
y = df['PlayTennis']

X.head(), y.head()
```

## Step 4: Train-Test Split

We split data into training and testing sets. Because this dataset is very small, accuracy can change depending on the split.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.30, random_state=42, stratify=y
)

print('Training rows:', X_train.shape[0])
print('Testing rows:', X_test.shape[0])
```

## Step 5: Build Naive Bayes Model

We use:
- `OrdinalEncoder` to convert text categories into numbers
- `CategoricalNB` because all features are categorical

```python
model = Pipeline([
    ('encoder', OrdinalEncoder()),
    ('naive_bayes', CategoricalNB())
])

model.fit(X_train, y_train)
```

## Step 6: Make Predictions

```python
y_pred = model.predict(X_test)

result = pd.DataFrame({
    'Actual': y_test.values,
    'Predicted': y_pred
})
result
```

## Step 7: Evaluate the Model

```python
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred))
```

## Step 8: Predict New Data

Example: Outlook = Sunny, Temperature = Hot, Humidity = High, Wind = Weak

```python
new_data = pd.DataFrame({
    'Outlook': ['Sunny'],
    'Temperature': ['Hot'],
    'Humidity': ['High'],
    'Wind': ['Weak']
})

prediction = model.predict(new_data)
probability = model.predict_proba(new_data)

print('Prediction:', prediction[0])
print('Class Labels:', model.named_steps['naive_bayes'].classes_)
print('Probabilities:', probability)
```

## Step 9: Another Prediction

Example: Outlook = Overcast, Temperature = Mild, Humidity = Normal, Wind = Weak

```python
new_data_2 = pd.DataFrame({
    'Outlook': ['Overcast'],
    'Temperature': ['Mild'],
    'Humidity': ['Normal'],
    'Wind': ['Weak']
})

print('Prediction:', model.predict(new_data_2)[0])
print('Probabilities:', model.predict_proba(new_data_2))
```
