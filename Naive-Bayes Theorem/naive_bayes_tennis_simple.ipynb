# Naive Bayes - Tennis Example (Simple)

This example predicts whether to play tennis based on weather conditions.

```python
import pandas as pd
from sklearn.naive_bayes import CategoricalNB
from sklearn.preprocessing import OrdinalEncoder
from sklearn.pipeline import Pipeline
```

## Step 1: Create Dataset

```python
data = {
    'Outlook': ['Sunny','Sunny','Overcast','Rain','Rain','Rain','Overcast','Sunny','Sunny','Rain','Sunny','Overcast','Overcast','Rain'],
    'Temperature': ['Hot','Hot','Hot','Mild','Cool','Cool','Mild','Hot','Cool','Mild','Mild','Mild','Hot','Mild'],
    'Humidity': ['High','High','High','High','Normal','Normal','Normal','High','Normal','Normal','Normal','High','Normal','High'],
    'Wind': ['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Weak','Weak','Strong','Strong','Weak','Strong'],
    'PlayTennis': ['No','No','Yes','Yes','Yes','No','Yes','No','Yes','Yes','Yes','Yes','Yes','No']
}

df = pd.DataFrame(data)
df
```

## Step 2: Split Features and Target

```python
X = df.drop('PlayTennis', axis=1)
y = df['PlayTennis']
```

## Step 3: Train Naive Bayes Model

```python
model = Pipeline([
    ('encoder', OrdinalEncoder()),
    ('nb', CategoricalNB())
])

model.fit(X, y)
```

## Step 4: Predict New Data

```python
test_data = pd.DataFrame({
    'Outlook': ['Sunny'],
    'Temperature': ['Hot'],
    'Humidity': ['High'],
    'Wind': ['Weak']
})

prediction = model.predict(test_data)
probability = model.predict_proba(test_data)

print('Prediction:', prediction[0])
print('Probabilities:', probability)
```

## Explanation

This model uses Naive Bayes:

P(Yes | Features) ∝ P(Yes) × P(Outlook|Yes) × P(Temperature|Yes) × ...

The class with highest probability is selected.
