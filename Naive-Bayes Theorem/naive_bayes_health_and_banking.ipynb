# Naive Bayes Theorem: Health Science + Banking
This notebook contains two practical Naive Bayes examples with datasets.

## Bayes Theorem
$$P(Class|Features)=\frac{P(Features|Class)\times P(Class)}{P(Features)}$$
Naive Bayes assumes features are independent given the class.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder
from sklearn.pipeline import Pipeline
from sklearn.naive_bayes import CategoricalNB
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

# Example 1: Health Science — Flu Risk Prediction
Predict **High Risk** or **Low Risk** from symptoms.

```python
health_df = pd.read_csv('naive_bayes_health_flu_dataset.csv')
health_df.head()
```

```python
health_df['flu_risk'].value_counts()
```

```python
X = health_df.drop('flu_risk', axis=1)
y = health_df['flu_risk']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42, stratify=y)
```

```python
model_health = Pipeline([
    ('encoder', OrdinalEncoder()),
    ('classifier', CategoricalNB())
])
model_health.fit(X_train, y_train)
y_pred = model_health.predict(X_test)
print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nConfusion Matrix:\n', confusion_matrix(y_test, y_pred))
print('\nClassification Report:\n', classification_report(y_test, y_pred))
```

```python
new_patient = pd.DataFrame({
    'fever': ['Yes'],
    'cough': ['Yes'],
    'body_pain': ['Yes'],
    'sore_throat': ['No'],
    'fatigue': ['Yes']
})
print('Prediction:', model_health.predict(new_patient)[0])
print('Class Labels:', model_health.named_steps['classifier'].classes_)
print('Probabilities:', model_health.predict_proba(new_patient))
```

# Example 2: Banking Industry — Loan Approval Prediction
Predict **Approved** or **Rejected** using customer profile.

```python
bank_df = pd.read_csv('naive_bayes_banking_loan_dataset.csv')
bank_df.head()
```

```python
bank_df['loan_status'].value_counts()
```

```python
X2 = bank_df.drop('loan_status', axis=1)
y2 = bank_df['loan_status']
X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.25, random_state=42, stratify=y2)
model_bank = Pipeline([
    ('encoder', OrdinalEncoder()),
    ('classifier', CategoricalNB())
])
model_bank.fit(X2_train, y2_train)
y2_pred = model_bank.predict(X2_test)
print('Accuracy:', accuracy_score(y2_test, y2_pred))
print('\nConfusion Matrix:\n', confusion_matrix(y2_test, y2_pred))
print('\nClassification Report:\n', classification_report(y2_test, y2_pred))
```

```python
new_applicant = pd.DataFrame({
    'income_level': ['Medium'],
    'credit_score': ['Good'],
    'employment_status': ['Employed'],
    'existing_debt': ['Low'],
    'previous_default': ['No']
})
print('Prediction:', model_bank.predict(new_applicant)[0])
print('Class Labels:', model_bank.named_steps['classifier'].classes_)
print('Probabilities:', model_bank.predict_proba(new_applicant))
```

## Teaching Summary
Health Science: symptoms → flu risk.
Banking: customer profile → loan approval.
Naive Bayes is fast, simple, and useful for categorical classification problems.
