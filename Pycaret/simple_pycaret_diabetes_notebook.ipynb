## 2. Import PyCaret Dataset Utility

PyCaret provides some built-in datasets for practice.

Here, we will use the **diabetes dataset**. The goal is to predict whether a person has diabetes or not.

```python
from pycaret.datasets import get_data

# Load diabetes dataset
data = get_data('diabetes')
```

## 3. Understand the Dataset

Before applying machine learning, we should always look at the dataset.

Important things to check:
- Number of rows and columns
- Column names
- Target/output column
- Missing values
- Data types

```python
# Show first five rows
data.head()
```

```python
# Shape of dataset
data.shape
```

```python
# Check column names
data.columns
```

```python
# Check missing values
data.isnull().sum()
```

## 4. Setup PyCaret Experiment

The `setup()` function prepares the complete machine learning environment.

It automatically handles many preprocessing steps such as:
- Train-test split
- Data type detection
- Missing value handling
- Feature preparation

In this dataset, the target column is **Class variable**.

```python
from pycaret.classification import *

# Setup classification experiment
clf_setup = setup(
    data=data,
    target='Class variable',
    session_id=42
)
```

## 5. Compare Different Machine Learning Models

The `compare_models()` function trains many models and compares their performance.

PyCaret shows a table with metrics such as:
- Accuracy
- AUC
- Recall
- Precision
- F1 Score

The best model is selected automatically.

```python
# Compare models and select the best one
best_model = compare_models()

best_model
```

## 6. Create a Specific Model

We can also create a specific model manually.

Here, we create a **Random Forest Classifier**. Random Forest is a popular machine learning algorithm that combines many decision trees.

```python
# Create Random Forest model
rf_model = create_model('rf')
```

## 7. Tune the Model

`tune_model()` improves the model by searching for better hyperparameters.

Hyperparameters are settings that control how the model learns.

```python
# Tune Random Forest model
tuned_rf = tune_model(rf_model)
```

## 8. Evaluate the Model

`evaluate_model()` opens an interactive panel where we can check:
- Confusion matrix
- ROC curve
- Feature importance
- Precision-recall curve

This helps us understand model performance visually.

```python
# Evaluate tuned model
evaluate_model(tuned_rf)
```

## 9. Plot Model Results

We can also generate individual plots using `plot_model()`.

The confusion matrix shows correct and incorrect predictions.

```python
# Confusion matrix
plot_model(tuned_rf, plot='confusion_matrix')
```

```python
# Feature importance
plot_model(tuned_rf, plot='feature')
```

## 10. Make Predictions

The `predict_model()` function tests the model on unseen/test data.

It adds prediction columns to the dataset.

```python
# Predict on test data
predictions = predict_model(tuned_rf)

predictions.head()
```

## 11. Finalize the Model

`finalize_model()` trains the selected model on the complete dataset.

This step is done before saving or deploying the model.

```python
# Finalize model
final_model = finalize_model(tuned_rf)

final_model
```

## 12. Save the Model

After training, we can save the model and use it later without training again.

```python
# Save model
save_model(final_model, 'diabetes_pycaret_model')
```

## 13. Load the Saved Model

We can load the saved model later and use it for predictions.

```python
# Load saved model
loaded_model = load_model('diabetes_pycaret_model')

loaded_model
```

## 14. Predict on New Data

To predict on new data, the new data must have the same feature columns as the original dataset, except the target column.

```python
# Example: take first 5 rows without target column
new_data = data.drop('Class variable', axis=1).head()

# Make predictions
new_predictions = predict_model(loaded_model, data=new_data)

new_predictions
```
